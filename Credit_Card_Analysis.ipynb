```python
df["Gender"] = df["Gender"].map({
    1: "Male",
    2: "Female"
})
```

```python
df["Education"] = df["Education"].map({
    0: "Unknown",
    1: "Graduate School",
    2: "University",
    3: "High School",
    4: "Others",
    5: "Unknown",
    6: "Unknown"
})
```

```python
df["Marital_Status"] = df["Marital_Status"].map({
    0: "Unknown",
    1: "Married",
    2: "Single",
    3: "Others"
})
```

```python
bill_columns = [
    "Bill_Sept",
    "Bill_Aug",
    "Bill_July",
    "Bill_June",
    "Bill_May",
    "Bill_April"
]

df["Total_Billed_6M"] = df[bill_columns].sum(axis=1)

print(df["Total_Billed_6M"])
```

```python
df["Avg_Monthly_Bill"] = df[bill_columns].mean(axis=1)
print(df["Avg_Monthly_Bill"])
```

```python
payment_columns = [
    "Payment_Sept",
    "Payment_Aug",
    "Payment_July",
    "Payment_June",
    "Payment_May",
    "Payment_April"
]

df["Total_Paid_6M"] = df[payment_columns].sum(axis=1)
print(df["Total_Paid_6M"])
```

```python
df["Avg_Monthly_Payment"] = (
    df[payment_columns].mean(axis=1)
)
print(df["Avg_Monthly_Payment"])
```

```python
df["Credit_Utilization"] = (
    df["Avg_Monthly_Bill"] /
    df["Credit_Limit"]
) * 100
print(df["Credit_Utilization"])
```

```python
pay_columns = [
    "Pay_Status_Sept",
    "Pay_Status_Aug",
    "Pay_Status_July",
    "Pay_Status_June",
    "Pay_Status_May",
    "Pay_Status_April"
]

df["Avg_Payment_Status"] = (
    df[pay_columns].mean(axis=1)
)
```

```python
def assign_segment(row):

    if row["Credit_Limit"] >= 200000 and row["Credit_Utilization"] < 50:
        return "Premium"

    elif row["Credit_Limit"] >= 100000 and row["Credit_Utilization"] < 75:
        return "High Value"

    elif row["Credit_Utilization"] >= 75:
        return "High Utilization"

    else:
        return "Standard"
```

```python
df["Customer_Segment"] = df.apply(
    assign_segment,
    axis=1
)
```

```python
contact_customers = np.random.choice(
    customers["Customer_ID"],
    size=num_contacts,
    replace=True
)
```

```python
channels = [
    "Email",
    "SMS",
    "Mobile App",
    "Phone",
    "Web"
]

campaign_contacts["Channel"] = np.random.choice(
    channels,
    size=num_contacts,
    p=[0.30, 0.20, 0.25, 0.10, 0.15]
)
```

```python
campaign_contacts["Response"] = np.random.choice(
    [0, 1],
    size=num_contacts,
    p=[0.75, 0.25]
)
```

```python
campaign_contacts["Conversion"] = np.where(
    campaign_contacts["Response"] == 1,
    np.random.choice([0, 1], size=num_contacts, p=[0.60, 0.40]),
    0
)
```

```python
print(
    campaign_contacts["Response"].value_counts()
)

print(
    campaign_contacts["Conversion"].value_counts()
)
```

```python
response_rate = campaign_contacts["Response"].mean()

print("Overall Response Rate:", response_rate)
```

```python
conversion_rate = (
    campaign_contacts["Conversion"].sum()
    / campaign_contacts["Response"].sum()
)

print("Conversion Rate:", conversion_rate)
```

```python
channel_cost = {
    "Email": 5,
    "SMS": 8,
    "Mobile App": 3,
    "Phone": 25,
    "Web": 2
}
campaign_contacts["Contact_Cost"] = (
    campaign_contacts["Channel"].map(channel_cost)
)
```

```python
converted_contacts = campaign_contacts[
    campaign_contacts["Conversion"] == 1
].copy()
print(converted_contacts.shape)
print(converted_contacts.head())
```

```python
campaign_transactions = transactions.merge(
    converted_contacts[
        [
            "Contact_ID",
            "Customer_ID",
            "Campaign_ID",
            "Channel"
        ]
    ],
    on="Customer_ID",
    how="inner"
)
```

```python
transactions["Transaction_Amount"]
```

```python
campaign_transactions["Revenue"] = (
    campaign_transactions["Transaction_Amount"]
)
```

```python
print(
    campaign_transactions[
        [
            "Customer_ID",
            "Campaign_ID",
            "Transaction_Amount",
            "Revenue"
        ]
    ].head()
)
```

```python
campaign_revenue = (
    campaign_transactions
    .groupby("Campaign_ID")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Revenue_Transactions=("Transaction_ID", "count"),
        Unique_Customers=("Customer_ID", "nunique")
    )
    .reset_index()
)
print(campaign_revenue)
```

```python
campaign_contacts["Contact_Cost"]
```

```python
campaign_cost = (
    campaign_contacts
    .groupby("Campaign_ID")
    .agg(
        Total_Cost=("Contact_Cost", "sum"),
        Total_Contacts=("Contact_ID", "count"),
        Total_Responses=("Response", "sum"),
        Total_Conversions=("Conversion", "sum")
    )
    .reset_index()
)
print(campaign_cost)
```

```python
campaign_performance = campaign_cost.merge(
    campaign_revenue,
    on="Campaign_ID",
    how="left"
)
```

```python
campaign_performance["Total_Revenue"] = (
    campaign_performance["Total_Revenue"].fillna(0)
)
```

```python
campaign_performance["Net_Revenue"] = (
    campaign_performance["Total_Revenue"]
    - campaign_performance["Total_Cost"]
)
```

```python
campaign_performance["ROI"] = (
    campaign_performance["Net_Revenue"]
    / campaign_performance["Total_Cost"]
) * 100
```

```python
campaign_performance["Response_Rate"] = (
    campaign_performance["Total_Responses"]
    / campaign_performance["Total_Contacts"]
) * 100
```

```python
campaign_performance["Conversion_Rate"] = (
    campaign_performance["Total_Conversions"]
    / campaign_performance["Total_Responses"]
).replace([np.inf, -np.inf], 0).fillna(0) * 100
```

```python
campaign_performance = campaign_performance.merge(
    campaigns[
        [
            "Campaign_ID",
            "Campaign_Name",
            "Campaign_Type",
            "Target_Segment"
        ]
    ],
    on="Campaign_ID",
    how="left"
)
```

```python
campaign_performance = campaign_performance[
    [
        "Campaign_ID",
        "Campaign_Name",
        "Campaign_Type",
        "Target_Segment",
        "Total_Contacts",
        "Total_Responses",
        "Response_Rate",
        "Total_Conversions",
        "Conversion_Rate",
        "Unique_Customers",
        "Total_Revenue",
        "Total_Cost",
        "Net_Revenue",
        "ROI"
    ]
]
print(campaign_performance)
```

```python
campaign_performance.to_csv(
    "campaign_performance.csv",
    index=False
)
```
